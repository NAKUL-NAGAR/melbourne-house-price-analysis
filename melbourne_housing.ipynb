```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:\\Users\\nakul\\Downloads\\archive (1)\\MELBOURNE_HOUSE_PRICES_LESS.csv")
```

```python
df.head()
```

```python
print(df.dtypes)
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["Price"]) # Remove rows with missing price values
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
# Remove duplicate rows


df = df.drop_duplicates()
```

```python
print(df["Price"].describe()) # Display summary statistics for the 'Price' column
```

```python
df["Price_per_Room"] = df["Price"] / df["Rooms"]  #Create Price per Room
print(df["Price_per_Room"])
```

```python
print("\nInvalid prices:")

print((df["Price"] <= 0).sum())


df = df[df["Price"] > 0]
```

```python
print(df["Suburb"].value_counts().head(10))  # Display the top 10 suburbs by count
```

```python
print(df.columns.tolist())
```

```python
print("\nInvalid rooms:")

print((df["Rooms"] <= 0).sum())


df = df[df["Rooms"] > 0]
```

```python
df["property_type"] = df["Type"].map({

    "h": "House",

    "u": "Unit",

    "t": "Townhouse"

})

print(df["property_type"].value_counts())
```

```python
print(df["Regionname"].value_counts())
```

```python
print(df.groupby("Rooms")["Price"].mean())
```

```python
print(df.duplicated().sum())
```

```python
print(df.shape)
```

```python
df.to_csv("melbourne_house_prices_cleaned.csv",index=False)
```
